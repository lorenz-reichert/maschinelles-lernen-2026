```python
import numpy as np
import math
import random
from sklearn.decomposition import PCA, FastICA, DictionaryLearning
from matplotlib import pyplot as plt
%matplotlib inline

def random_patches(data, img_height, img_width, n_patches, patch_size):
    # RANDOMPATCHES draw random square-shaped patches from the data.
    #
    # INPUT:
    #   data : ndarray containing the data (i.e. mnist or olivetti)
    #   img_height : height of the images
    #   img_width : width of the images
    #   n_patches : number of patches to be drawn
    #   patch_size : size (both width and height) of the square-shaped patches
    #
    # OUTPUT:
    #   patches : random patches

    # reshape data 2d -> 3d
    n_images = data.shape[0]
    data = data.reshape(n_images, img_height, img_width)

    # extract random patches
    patches = []#np.zeros((n_patches, patch_size*patch_size))
    for i in range(n_patches):
        patch_top = random.randint(0, img_height - patch_size)
        patch_left = random.randint(0, img_width - patch_size)
        p = data[random.randint(0, n_images-1), patch_top:patch_top+patch_size, patch_left:patch_left+patch_size]
        patches.append(p)
        #patches[i,:] = p[:]

    patches = np.stack(patches)
    patches = patches.reshape(patches.shape[0], -1) #flatten each patch

    return patches

def show_in_grid(images, height, width):
    # flatten images if necessary
    images = images.reshape(images.shape[0], -1)

    # normalize patches
    images = images / (np.abs(images).max(axis=1, keepdims=True)+1e-6)

    # reshape into images
    images = images.reshape(-1, height, width)

    # make images fit into a rectangular area
    grid_width = math.ceil(math.sqrt(images.shape[0]))
    grid_height = math.ceil(images.shape[0] / grid_width)

    empty_cells = grid_width * grid_height - images.shape[0]

    # fill empty cells
    if empty_cells > 0:
        padding = np.zeros((empty_cells, height, width))
        images = np.concatenate((images, padding))

    # rearrange basis into grid and also switch width and height so x and y axis are not switched
    images = images.reshape(grid_height, grid_width, height, width)
    images = images.transpose(0, 3, 1, 2)

    plt.figure(figsize = (10,10))
    plt.imshow(images.reshape(grid_height*height, grid_width*width), cmap='gray')
    plt.show()
```

#### PCA
The PCA is defined as $\vec{x}_{PCA}=U^T\vec{x}$ (equation 1.2) where the matrix $U$ is the matrix that diagonalizes the covariance matrix $C=\langle\vec{x}_{i}\vec{x}_{i}^T\rangle$.

#### Whitening
The transformation $\vec{x}_{wht}=S^{-1}U^T\vec{x}$ (equation 1.3) then performs the so-called whitening and we obtain $\langle\vec{x}_{wht}\vec{x}_{wht}^T\rangle=S^{-1}U^TCUS^{-1}=1$.

**Answer:** False. The optimal number of principal components to retain is not necessarily 42. Optimal number of components depends on the dataset and on what one desires to achieve for the trade-off between reduction of dimensions and information loss/preservation:  reduce dimensions => less components => information loss // preserve information => more components => more dimensions

4. After whitening, the features of an observation are uncorrelated and have unit variance.

**Answer:** Correct. After whitening, the covariance matrix is the unit matrix with covariance=0 (features uncorrelated) and variance=1

5. The resulting basis vectors of the ICA (i.e. the independent components) are mutually orthogonal.

**Answer:** False. PCA results in orthogonal basis vectors, but ICA doesn't. Instead, it aims for statistical independence among the components, which does not mean that the basis vectors are mutually orthogonal.

6. The ICA is an enhancement of the PCA which additionally aims to make the already uncorrelated features statistically independent.

**Answer:** Yes, this is correct. ICA aims for statistical independency, PCA only makes components uncorrelated

7. The set of basis vectors of the ICA can be over-complete, i.e. more basis vectors than dimensions of the input space.

 **Answer:** False. Basis vectors of ICA have same number of components as input dimension. However, Sparse Coding (SC) may result in an over-complete set of basis vectors with more basis vectors than input dimensions

8. Sparsity means that only a few basis vectors are used to reconstruct an image.

**Answer:** Correct. Only few coefficients are non-zero and as a result only few basis vectors are used to reconstruct an image, the rest of the coefficients are (set to) zero.

9. The optimization problem for SC (c.f. equation 1.9 in the lecture notes) is convex.

**Answer:** False. According to the script, SC optimizes 1) $\vec{x}_{SC}$ in the cost function $E$ (the coefficients, which means it optimizes the "intensity" with which the basis vectors are used) and 2) the rows of $W^{-1}$ which are the basis vectors themselves. The optimization problem is formulated as $\min_{W^{-1}}(\min_{\vec{x}_{SC}}{E})$. This is a nested optimization problem and it is non-convex as it is possible to have local minima that are not global minima.
